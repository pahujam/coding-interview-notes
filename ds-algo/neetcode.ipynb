#### **Maximum Subarray Sum (Kadane's Algorithm)**


- Imagine we have array of daily stock prices movement = nums
- Maximum Profit = maximum subarray sum of nums array ie. what continuous block of the days we can hold the stock to give us max sum of daily changes
- We add daily changes continuously, tracking the maximum sum and the current sum. If current sum becomes negative, we reset the current sum to zero, (assuming we buy the stock when it bottoms out) and continue rest

```python
def maxSubArray(nums):
  maxSub = nums[0]
  curSum = 0

  for n in nums:
    if curSum < 0:
      curSum = 0
    curSum += n
    maxSub = max(maxSub, curSum)

  return maxSub

nums = [-9,4,1,1,-6]
maxSubArray(nums)
```

#### **Best Time to Buy and Sell Stock**
- same as maximum subarray sum
- use kadane's algorithm to solve this

```python
def maxProfit(prices) -> int:
  if len(prices)==1:
      return 0

  diffs = []
  for i in range(len(prices)-1):
      diffs.append(prices[i+1] - prices[i])

  maxSub = diffs[0]
  curSum = 0

  for n in diffs:
      if curSum < 0:
          curSum = 0
      curSum += n
      maxSub = max(maxSub, curSum)

  return max(0,maxSub)

prices = [7, 6, 4, 3, 1]
maxProfit(prices)
```

#### **Valid Palindrome**
- string reads the same from left and right
- solution: two pointer approach

```python
def is_palindrome(str1):
  left = 0
  right = len(str1)-1

  while left < right:
    if str1[left] != str1[right]:
      return False
    else:
      left += 1
      right -= 1

  return True
```

#### **Valid Anagram**
- one solution = compare sorted strings
- other solution = build hash tables and compare frequencies

```python
def is_anagram(str1, str2):
  if len(str1)!=len(str2):
    return False
  count_str1, count_str2 = {}, {}

  for i in range(len(str1)):
    count_str1[str1[i]] = 1 + count_str1.get(str1[i],0)
    count_str2[str2[i]] = 1 + count_str2.get(str2[i],0)

  return count_str1 == count_str2
```

#### **Kth Largest Element in a Stream**

```python
import heapq
class KthLargest:
  def __init__(self, k, nums):
    self.minHeap = nums
    self.k = k
    heapq.heapify(self.minHeap) # O(n) runtime

    while len(self.minHeap) > k:
      heapq.heappop(self.minHeap)

  def add(self, val):
    heapq.heappush(self.minHeap, val)

    if len(self.minHeap) > self.k:
      heapq.heappop(self.minHeap)
    return self.minHeap[0]
```

#### **Longest Substring Without Repeating Characters**
- use the sliding window approach

```python
def longest_substr(str1):
  char_set = set()
  l = 0
  res = 0

  for r in range(len(str1)):
    while str1[r] in char_set:
      char_set.remove(str1[l])
      l += 1
    char_set.add(str1[r])
    res = max(res, r - l + 1)
  return res
```

#### **Min Stack**
- we maintain two stacks: one normal + one to store minimums at each element position

```python
class MinStack:
  def __init__(self):
    self.stack = []
    self.minStack = []

  def push(self, val: int) -> None:
    self.stack.append(val)
    val = min(val, self.minStack[-1] if self.minStack else val)
    self.minStack.append(val)

  def pop(self) -> None:
    self.stack.pop()
    self.minStack.pop()

  def top(self) -> int:
      return self.stack[-1]

  def getMin(self) -> int:
      return self.minStack[-1]
```

#### **Queue using Stacks**

```python
class MyQueue:
  def __init__(self):
    self.in_stack = []
    self.out_stack = []

  def push(self, x):
    # O(1)
    self.in_stack.append(x)

  def peek(self):
    # O(1) amortized
    if not self.out_stack:
      while self.in_stack:
        self.out_stack.append(self.in_stack.pop())

    return self.out_stack[-1]

  def pop(self):
    # O(1) amortized
    self.peek()
    return self.out_stack.pop()

  def empty(self):
    return not self.in_stack and not self.out_stack
```

#### **Stack using Queues**
- Single Queue Strategy: every time we push new element, we rotate the elements by dequeue + enqueue to maintain stack like behaviour

```python
from collections import deque

class Stack:
  def __init__(self):
    self.q = deque()

  def push(self, x):
    size = len(self.q)

    # enqeue x
    self.q.append(x)

    # element rotation
    for _ in range(size):
      self.q.append(self.q.popleft())

  def empty(self):
    return len(self.q) == 0

  def pop(self):
    if self.empty():
      raise IndexError("Empty Stack")
    return self.q.popleft()

  def top(self):
    if self.empty():
      raise IndexError("Empty Stack")
    return self.q[0]
```

#### **Two Sum**

```python
nums = [3,4,5,6]
target = 7

def two_sum(nums, target):
  diffs_dict = {}
  for i, n in enumerate(nums):
    if n in diffs_dict.keys():
      return [diffs_dict[n],i]
    else:
      diff = target - n
      diffs_dict[diff] = i

two_sum(nums, target)
```

#### **Reverse a Linked List**

```python
class ListNode:
  def __init__(self, val):
    self.val = val
    self.next = None
```

```python
def reverseList(head):
  curr = head
  prev = None

  while curr:

    # temp store the next node
    next_node = curr.next

    # reverse the current node pointer
    curr.next = prev

    # move prev and curr forward
    prev = curr
    curr = next_node

  # prev is the new head of reversed list
  return prev
```

#### **Valid Parentheses**

```python
def is_valid(str1):
  bracket_map = {")": "(", "}": "{", "]": "["}
  stack = []

  for char in str1:
    if char in bracket_map.keys():
      # pop dummy value if stack is empty, must return False
      top_elem = stack.pop() if stack else "#"

      if bracket_map[char] != top_elem:
        return False
    else:
      stack.append(char)

  # stack must be empty, else brackets were not closed
  return not stack
```
