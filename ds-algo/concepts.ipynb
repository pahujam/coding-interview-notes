### **DP**
**optimal Substructure**
- solution to larger problem is combination of solution to smaller subproblems.

**overlapping Subproblems**
- these smaller subproblems are having overlapping computations.

**Memoization**
- This is top-down DP Approach. Here we cache the results as soon as we compute them

````python
def fibo(n):
  memo = {}
  if n in memo:
    return memo[n]
  else:
    if n <= 1:
      memo[n] = n
    else:
      memo[n] = fibo(n-1) + fibo(n-2)
    return memo[n]

print(fibo(10))
````

**Tabluation**
- This is bottom-up DP Approach. Here we fill the array by computing results in bottom up fashion

````python
def fibo(n):
  F = [0] * (n+1)
  F[0] = 0
  F[1] = 1

  for i in range(2,n+1):
    F[i] = F[i-1] + F[i-2]

  return F[n]

print(fibo(10))
````

### **Sorting**

**Bubble Sort**
- Basic Implementation

````python
my_array = [-10, -20, 11, 12, 64, 34, 25, 12, 22, 11, 90, 5]
n = len(my_array)

for j in range(0, n-1):
  for i in range(0, n-1):
    if my_array[i+1] < my_array[i]:
      my_array[i], my_array[i+1] = my_array[i+1], my_array[i]

print(my_array)
````

**Optimization:**
- Last elem becomes biggest after 1st loop.
- 2nd Last elem becomes biggest after 2nd loop.

````python
my_array = [-10, -20, 11, 12, 64, 34, 25, 12, 22, 11, 90, 5]
n = len(my_array)

for j in range(0, n-1):
  for i in range(0, n-j-1):
    if my_array[i+1] < my_array[i]:
      my_array[i], my_array[i+1] = my_array[i+1], my_array[i]

print(my_array)
````

**Optimization:**
- If array is almost sorted, can save time by counting number of swaps
- If no swaps, means sorting is complete

````python
my_array = [-10, -20, 11, 12, 64, 34, 25, 12, 22, 11, 90, 5]
n = len(my_array)

for j in range(0, n-1):
  swapped = False
  for i in range(0, n-j-1):
    if my_array[i+1] < my_array[i]:
      my_array[i], my_array[i+1] = my_array[i+1], my_array[i]
      swapped = True
  if not swapped:
    break

print(my_array)
````

**Selection Sort**
- Pick the min element and move it to the correct position

````python
my_array = [-10, -20, 11, 12, 64, 34, 25, 12, 22, 11, 90, 5]
n = len(my_array)

for i in range(n-1):
  min_index = i
  for j in range(i+1,n):
    if my_array[j] < my_array[min_index]:
      min_index = j
  min_value = my_array.pop(min_index)
  my_array.insert(i, min_value)

print(my_array)
````

- Above code calls pop and insert functions on the list, hence requires shifting operations.
- Optimize this algorithm by just swapping the values

````python
my_array = [-10, -20, 11, 12, 64, 34, 25, 12, 22, 11, 90, 5]
n = len(my_array)

for i in range(n-1):
  min_index = i
  for j in range(i+1,n):
    if my_array[j] < my_array[min_index]:
      min_index = j
  my_array[i], my_array[min_index] = my_array[min_index], my_array[i]

print(my_array)
````

**Merge Sort**

````python
def mergeSort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    leftHalf = arr[:mid]
    rightHalf = arr[mid:]

    sortedLeft = mergeSort(leftHalf)
    sortedRight = mergeSort(rightHalf)

    return merge(sortedLeft, sortedRight)

def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result

my_array = [-10, -20, 11, 12, 64, 34, 25, 12, 22, 11, 90, 5]
print(mergeSort(my_array))
````

**Merge Two Sorted Arrays**

````python
def merge_arrays(arr1, arr2):
    merged = []
    i, j = 0, 0

    # Traverse both arrays
    while i < len(arr1) and j < len(arr2):
        if arr1[i] < arr2[j]:
            merged.append(arr1[i])
            i += 1
        else:
            merged.append(arr2[j])
            j += 1

    # Append remaining elements of arr1 (if any)
    while i < len(arr1):
        merged.append(arr1[i])
        i += 1

    # Append remaining elements of arr2 (if any)
    while j < len(arr2):
        merged.append(arr2[j])
        j += 1

    return merged

# Example usage:
arr1 = [1, 3, 5]
arr2 = [2, 4, 6]
result = merge_arrays(arr1, arr2)
print(result)  # Output: [1, 2, 3, 4, 5, 6]
````

### **Dictionary**
- Dictionary is implemented using Hash Table

````python
class MyDict:
    def __init__(self):
        self.size = 10  # Initial size of the hash table
        self.table = [[] for _ in range(self.size)]  # List of lists for chaining

    def _hash(self, key):
        return hash(key) % self.size

    def set(self, key, value):
        index = self._hash(key)
        for i, (k, v) in enumerate(self.table[index]):
            if k == key:
                self.table[index][i] = (key, value)  # Update existing key
                return
        self.table[index].append((key, value))  # Insert new key-value pair

    def get(self, key):
        index = self._hash(key)
        for k, v in self.table[index]:
            if k == key:
                return v
        return None  # Key not found

    def remove(self, key):
        index = self._hash(key)
        for i, (k, v) in enumerate(self.table[index]):
            if k == key:
                del self.table[index][i]  # Remove key-value pair
                return True
        return False  # Key not found

    def __str__(self):
        return str({k: v for bucket in self.table for k, v in bucket})

# Example usage:
my_dict = MyDict()
my_dict.set("apple", 1)
my_dict.set("banana", 2)
print(my_dict)  # Output: {'apple': 1, 'banana': 2}
print(my_dict.get("apple"))  # Output: 1
my_dict.remove("banana")
print(my_dict)  # Output: {'apple': 1}

# Notes:
# Runtimes of dict ops:
# 1. insert/update/get/remove -> O(1) avg case assuming good hash distribution, O(n) worst case
````

### **Set**
- Set is implemented using Hash Table

````python
class MySet:
    def __init__(self):
        self.size = 10  # Initial size of the hash table
        self.table = [[] for _ in range(self.size)]  # List of lists for chaining

    def _hash(self, key):
        return hash(key) % self.size

    def add(self, key):
        index = self._hash(key)
        for item in self.table[index]:
            if item == key:
                return  # Key already exists, do nothing
        self.table[index].append(key)  # Add the key

    def remove(self, key):
        index = self._hash(key)
        for i, item in enumerate(self.table[index]):
            if item == key:
                del self.table[index][i]  # Remove the key
                return True
        return False  # Key not found

    def contains(self, key):
        index = self._hash(key)
        for item in self.table[index]:
            if item == key:
                return True
        return False  # Key not found

    def __str__(self):
        return str({item for bucket in self.table for item in bucket})

# Example usage:
my_set = MySet()
my_set.add(1)
my_set.add(2)
print(my_set)  # Output: {1, 2}
print(my_set.contains(1))  # Output: True
my_set.remove(2)
print(my_set)  # Output: {1}

# Runtimes:
# add/remove/contains = O(1) avg time, O(n) worst case (in case of my collisions)
````

### **Stack**
- Stack is implemented using list

````python
class Stack:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        else:
            raise IndexError("Pop from an empty stack")

    def peek(self):
        if not self.is_empty():
            return self.items[-1]
        else:
            raise IndexError("Peek from an empty stack")

    def size(self):
        return len(self.items)

    def __str__(self):
        return str(self.items)

# Example usage:
stack = Stack()
stack.push(1)
stack.push(2)
stack.push(3)
print(stack)  # Output: [1, 2, 3]
print(stack.pop())  # Output: 3
print(stack.peek())  # Output: 2
print(stack.size())  # Output: 2

# Notes:
# Runtimes:
# push/pop/peek/size -> O(1) time
````

### **Queues**

**Queue Using LinkedList**
```
head  →  ...  →  tail
^ dequeue        enqueue ^
```



````python
class Node:
  def __init__(self, v):
        self.v = v
        self.next = None

class Queue:
  def __init__(self):
    self.head = None
    self.tail = None

  def enqueue(self, x):
    n = Node(x)
    if not self.tail:
      self.head = self.tail = n
    else:
      self.tail.next = n
      self.tail = n

  def dequeue(self):
    if not self.head:
      return None
    v = self.head.v
    self.head = self.head.next
    if self.head is None:
      self.tail = None
    return v
````

**Deque Class**

````python
from collections import deque

# create a queue from a list
d = deque([1, 2, 3])
print(d)

d.append(4)        # right end
d.appendleft(0)    # left end
print(d)

d.pop()            # remove right
d.popleft()        # remove left
print(d)

print(d[0])       # leftmost
print(d[-1])      # rightmost

d.rotate(1)   # move right end to left
print(d)
````

**DFS: Pre-Order Traversal**

````python
def dfs_preorder(node):
  if node is None:
    return None
  print(node.data, end=",")
  dfs_preorder(node.left)
  dfs_preorder(node.right)

dfs_preorder(root)
````

````python
def dfs_postorder(node):
  if node is None:
    return None

  dfs_preorder(node.left)
  dfs_preorder(node.right)
  print(node.data, end=",")

dfs_postorder(root)
````

````python
def dfs_inorder(node):
  if node is None:
    return None

  dfs_preorder(node.left)
  print(node.data, end=",")
  dfs_preorder(node.right)

dfs_inorder(root)
````

**Binary Tree Implementation with Array**
- If we read from binary tree a lot, instead of modifying it, Array Implementation will make more sense
- Left Child is stored in `2*i + 1`
- Right Child is stored in `2*i + 2`

````python
binary_tree_array = ['R', 'A', 'B', 'C', 'D', 'E', 'F', None, None, None, None, None, None, 'G']

def left_child_index(index):
    return 2 * index + 1

def right_child_index(index):
    return 2 * index + 2

def pre_order(index):
    if index >= len(binary_tree_array) or binary_tree_array[index] is None:
        return []
    return [binary_tree_array[index]] + pre_order(left_child_index(index)) + pre_order(right_child_index(index))

def in_order(index):
    if index >= len(binary_tree_array) or binary_tree_array[index] is None:
        return []
    return in_order(left_child_index(index)) + [binary_tree_array[index]] + in_order(right_child_index(index))

def post_order(index):
    if index >= len(binary_tree_array) or binary_tree_array[index] is None:
        return []
    return post_order(left_child_index(index)) + post_order(right_child_index(index)) + [binary_tree_array[index]]

print("Pre-order Traversal:", pre_order(0))
print("In-order Traversal:", in_order(0))
print("Post-order Traversal:", post_order(0))
````

### **BST**

**Advantages:**
 - Search, Insert, Delete are O(logn) operations

**Properties of BST:**
  - Any node's left child and all of its descendants must have lower value than the node itself.
  - Any node's right child and all of its descendants must have higher value than the node itself.
  - Left and right subtrees must also be BST

**Terms and Meanings:**
  - size = num of nodes
  - node's height = maximum number of edges b/w node and a leaf node
  - node's in-order successor = node that comes after if we do an in-order traversal
  - to check if the Tree is BST, do an in-order traversal and the result must be in increasing order.

````python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def inOrderTraversal(node):
    if node is None:
        return
    inOrderTraversal(node.left)
    print(node.data, end=", ")
    inOrderTraversal(node.right)

root = TreeNode(13)
node7 = TreeNode(7)
node15 = TreeNode(15)
node3 = TreeNode(3)
node8 = TreeNode(8)
node14 = TreeNode(14)
node19 = TreeNode(19)
node18 = TreeNode(18)

root.left = node7
root.right = node15

node7.left = node3
node7.right = node8

node15.left = node14
node15.right = node19

node19.left = node18

print(inOrderTraversal(root))
````

**Search in BST**
- Time complexity is O(h)

````python
def search(node, target):
  if node is None:
    return None
  elif node.data == target:
    return node
  elif target < node.data:
    return search(node.left, target)
  else:
    return search(node.right, target)

print(search(root, 15).data)
````

**Insert in BST**

````python
def insert(node, data):
  if node is None:
    return TreeNode(data)
  else:
    if data < node.data:
      node.left = insert(node.left, data)
    elif data > node.data:
      node.right = insert(node.right, data)
    return node

print(inOrderTraversal(root))
insert(root, 16)
print(inOrderTraversal(root))
````

**Find Min Value**

````python
def minValueNode(node):
    current = node
    while current.left is not None:
        current = current.left
    return current

print(minValueNode(root).data)
````

**Delete a Node in BST**
- If node has no child, delete it directly.
- If node has 1 child, attach the child to the parent of deleted node
- If node has 2 child, replace in-order successor of deleted node to the parent node and delete the replaced in-order node

````python
def delete(node, data):
  if node is None:
    return None
  if data < node.data:
    node.left = delete(node.left, data)
  elif data > node.data:
    node.right = delete(node.right, data)
  else:
    # Node with only one child or no child
    if not node.left:
      temp = node.right
      return temp
    elif not node.right:
      temp = node.left
      return temp

    # Node with two children, get the in-order successor
    node.data = minValueNode(node.right).data
    node.right = delete(node.right, node.data)
  return node
````

### **Backtracking**

**Find all subsets of nums**

````python
nums = [1,2,5]
subset = []
results = []

def backtrack(i, subset):
    print(i, subset)
    if i == len(nums):
        results.append(subset.copy())
        return None

    # if i is included
    subset.append(nums[i])
    backtrack(i+1, subset)

    # if i is not included
    subset.pop()
    backtrack(i+1, subset)

backtrack(0, subset)
print("results: ", results)
````

### **Math**

**Check Prime**

- A prime number is a natural number (positive integer) > 1 with only two distinct positive divisors: 1 and itself.

````python
def is_prime(n):
    if n < 2: return False
    i = 2
    while i*i <= n:
        if n % i == 0: return False
        i += 1
    return True

print(is_prime(29))
````

**Find Greatest Common Divisor**
- Use Euclidean Algorithm
```
Key fact (core identity)

For a = b*q + r (with 0 ≤ r < |b|),
every common divisor of a and b also divides r, and conversely any divisor of b and r divides a.
Therefore
gcd(a, b) = gcd(b, r).

This identity lets us replace the pair (a,b) by the smaller pair (b,r) without changing the gcd.
```

````python
def gcd(a,b):
  a, b = abs(a), abs(b)
  while b:
    a, b = b, a % b
  return a

print(gcd(30,20))
print(gcd(24,16))
````

- For LCM, use identity: a x b = lcm x gcd

````python
def lcm(a, b):
    return abs(a*b) // gcd(a, b)
````

**Linked List Cycle Detection**
- Use Floyd’s tortoise-and-hare cycle detection algorithm.

````python
def hasCycle(head):
    if not head or not head.next:
        return False

    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True
    return False

print(hasCycle(singly_ll))
print(hasCycle(doubly_ll))
print(hasCycle(circular_ll))
````

### **Heaps**
- A complete binary tree is a binary tree where all levels are fully filled except possibly the last, and the last level’s nodes are placed left to right with no gaps.
- A heap is a binary tree–based data structure with two rules:
  - It’s always a complete binary tree (filled left-to-right)
  - Max-heap: Every node ≥ its children.
- Because of this, the root always holds the minimum (min-heap) or maximum (max-heap) value.

**Heap operation runtimes** (min-heap or max-heap):
- Get top (peek): O(1)
- Insert (push): O(log n)
- Remove top (pop): O(log n)
- Search for arbitrary value: O(n)
- Build heap from array: (Two ways)
  - Repeated insert: O(n log n)
  - Bottom-up heapify: O(n)












**Max Heap Implementation**

````python
class MaxHeap:
    def __init__(self):
        self.h = []

    def parent(self, i): return (i - 1) // 2
    def left(self, i): return 2*i + 1
    def right(self, i): return 2*i + 2

    def push(self, x):
        self.h.append(x)
        i = len(self.h) - 1
        while i > 0 and self.h[i] > self.h[self.parent(i)]:
            p = self.parent(i)
            self.h[i], self.h[p] = self.h[p], self.h[i]
            i = p

    def pop(self):
        if not self.h: return None
        root = self.h[0]
        last = self.h.pop()
        if self.h:
            self.h[0] = last
            self._heapify(0)
        return root

    def _heapify(self, i):
        largest = i
        l, r = self.left(i), self.right(i)
        if l < len(self.h) and self.h[l] > self.h[largest]:
            largest = l
        if r < len(self.h) and self.h[r] > self.h[largest]:
            largest = r
        if largest != i:
            self.h[i], self.h[largest] = self.h[largest], self.h[i]
            self._heapify(largest)

    def build_heap(self, arr):
        self.h = arr[:]
        n = len(self.h)
        for i in range(n//2 - 1, -1, -1):
            self._heapify(i)

# example
arr = [3, 1, 6, 5, 2, 4]
hp = MaxHeap()
hp.build_heap(arr)
print(hp.h)  # [6, 5, 4, 3, 2, 1]

hp.push(-1)
print(hp.pop())
print(hp.h)
````

**Heapq Class**

````python
import heapq

# create empty min-heap
h = []

# insert (push)
heapq.heappush(h, 5)
heapq.heappush(h, 1)
heapq.heappush(h, 3)
print(h)

# remove smallest
small = heapq.heappop(h)
print(h, small)

# peek smallest
top = h[0]

# turn list into heap in-place
arr = [4, 2, 7, 1]
heapq.heapify(arr)
print(arr)

# max-heap trick (store negatives)
mh = []
heapq.heappush(mh, -10)
heapq.heappush(mh, -3)
print(mh)

max_val = -heapq.heappop(mh)
print(mh, max_val)

# priority queue: (priority, item)
pq = []
heapq.heappush(pq, (2, "B"))
heapq.heappush(pq, (1, "A"))
heapq.heappush(pq, (3, "C"))
print(pq)

prio, item = heapq.heappop(pq)
print(pq, prio, item)
````

**Heap Sort**

````python
import heapq

my_array = [-10, -20, 11, 12, 64, 34, 25, 12, 22, 11, 90, 5]
out = []

h = my_array[:]
heapq.heapify(h) # O(n)
while h: # n times
  min_elem = heapq.heappop(h) # O(logn)
  out.append(min_elem)

print(out) # runtime for sorting is O(nlogn)
````

### **Graphs**
- **Weighted**: edges have weights.
- **Connected**: all verticies are connected through edges somehow. ie. no isolated (disjoint) subgraphs or single isolated vertices.
- **Directed**: when edges have a direction.
- **Cyclic**: when graphs have cycles. can be directed or undirected both.
- **Loop**: also called self-loop, is an edge that starts and ends on the same node/vertex.


**Adjacency Matrix Implementation**

````python
class Graph:
  def __init__(self, vertices):
    self.idx = {v:i for i,v in enumerate(vertices)}
    n = len(vertices)
    self.mat = [[0]*n for _ in range(n)]

  def add_edge(self, a, b):
    i, j = self.idx[a], self.idx[b]
    self.mat[i][j] = 1 # change to w for weighted graph
    self.mat[j][i] = 1 # remove this for directed graph

  def remove_edge(self, a, b):
    i, j = self.idx[a], self.idx[b]
    self.mat[i][j] = 0
    self.mat[j][i] = 0 # remove this for directed graph

verts = ["A", "B", "C", "D"]
g = Graph(verts)

g.add_edge("A", "C")
g.add_edge("B", "D")

print(g.mat)        # full matrix
print(g.idx["C"])   # index of a vertex
````

**Adjacency List Implementation**

````python
class Graph:
    def __init__(self, vertices):
        self.adj = {v: [] for v in vertices}

    def add_edge(self, a, b):
        self.adj[a].append(b)
        self.adj[b].append(a)

    def remove_edge(self, a, b):
        self.adj[a].remove(b)
        self.adj[b].remove(a)

# example
g = Graph(["A","B","C","D"])
g.add_edge("A","C")
g.add_edge("B","D")
print(g.adj)
````

**Ajdacency Matrix with BFS and DFS**

````python
class Graph:
    def __init__(self, vertices):
        self.idx = {v:i for i,v in enumerate(vertices)}
        self.verts = vertices
        n = len(vertices)
        self.mat = [[0]*n for _ in range(n)]

    def add_edge(self, a, b):
        i, j = self.idx[a], self.idx[b]
        self.mat[i][j] = 1
        self.mat[j][i] = 1

    def remove_edge(self, a, b):
        i, j = self.idx[a], self.idx[b]
        self.mat[i][j] = 0
        self.mat[j][i] = 0

    def neighbors(self, v):
        i = self.idx[v]
        return [self.verts[j] for j,v2 in enumerate(self.mat[i]) if v2]

    def bfs(self, start):
        from collections import deque
        seen = set([start])
        q = deque([start])
        order = []
        while q:
            v = q.popleft()
            order.append(v)
            for n in self.neighbors(v):
                if n not in seen:
                    seen.add(n); q.append(n)
        return order

    def dfs(self, start):
        seen = set()
        order = []
        def go(v):
            seen.add(v); order.append(v)
            for n in self.neighbors(v):
                if n not in seen:
                    go(n)
        go(start)
        return order


verts = ["A", "B", "C", "D"]
g = Graph(verts)
g.add_edge("A","C")
g.add_edge("B","D")
g.add_edge("A","B")

print(g.bfs("A"))
print(g.dfs("A"))
````

**Adjacency Matrix with Cycle Detection**
- Do DFS on graph, if same node is visited twice, it means it has a cycle
- Do DFS on each node, to complete check all subgraphs


````python
class Graph:
  def __init__(self, vertices):
    self.idx = {v:i for i,v in enumerate(vertices)}
    self.verts = vertices
    n = len(vertices)
    self.mat = [[0]*n for _ in range(n)]

  def add_edge(self, a, b):
    i, j = self.idx[a], self.idx[b]
    self.mat[i][j] = 1
    self.mat[j][i] = 1

  def neighbors(self, v):
    i = self.idx[v]
    return [self.verts[j] for j,x in enumerate(self.mat[i]) if x]

  def has_cycle(self):
    seen = set()
    def go(v, parent):
        seen.add(v)
        for n in self.neighbors(v):
            if n not in seen:
                if go(n, v): return True
            elif n != parent:
                return True
        return False
    for v in self.verts:
        if v not in seen and go(v, None):
            return True
    return False

g = Graph(["A","B","C"])
g.add_edge("A","B")
print(g.has_cycle())
g.add_edge("B","C")
print(g.has_cycle())
g.add_edge("C","A")
print(g.has_cycle())
````
