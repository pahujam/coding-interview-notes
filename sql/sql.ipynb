- Window fns: find rank of a column!
- row number = in case of tie, assign different ranks randomly!
- rank = assign equal rank if tie, but create a gap!
- dense rank = assign equal rank if tie, but no gaps

````python
%%sql
select *,
row_number() over(order by salary) as salary_rank
from employee
order by salary asc
````

````python
%%sql
select *,
rank() over(order by salary) as salary_rank
from employee
order by salary asc
````

### Different Types of Joins in SQL
 - Cross Join: Get the cartesian product of all rows


````python
%%sql
select * from department cross join employee
````

- Join / Inner Join: Filter rows from cartesian product only based on the matching condition
- The word inner is optional

````python
%%sql
select * from employee join department on employee.departmentId = department.id
````

````python
%%sql
select * from employee inner join department on employee.departmentId = department.id
````

- Left/Right Join: Keep non matching records from Left/Right Table

````python
%%sql
select * from employee left join department on employee.departmentId = department.id
````

#### FULL OUTER JOIN
- Keep all non matching records from both the tables.
- Returns rows even if no matching rows are found in other table.
- Same as outer Join.

````python
%%sql
select * from employee full outer join department on employee.departmentId = department.id
-- there is error bcoz mysql doesnt support full outer join, but syntax remains same
````

#### Semi Join
- Keep Rows from left table only when there is a matching row in the right table.
- Case 1: No match in right table -> Drop the row
- Case 2: Match in right table -> Keep the row
- Case 3: Multiple matches in right table -> Keep only 1 time (ie. dont repeat)
- Semi Join acts as a filter in Left Table based on keys in Right Table

#### Anti Join
- Keep Rows from Left Table only when there is **NO** matching row in Right Table.
- ie. Semi Join + Anti Join = Left Join

#### Natural Join
- Join tables based on keys with same column name without using the on condition
- Drawback: Bad because its not explicit about the keys based on which join is happening


#### Cross Join
- Computes the cartesian product of two tables.
- Does not have an ON Condition for join to happen.


#### Using Keyword
- Joins 2 tables based on Key name that is same in both the tables

````python
%%sql
with department_1 as (
  select id as departmentId,name from department
)
select * from department_1 join employee using (departmentId)
````

#### Lead / Lag Window Functions
- Syntax:


```
 LAG(column, offset, default_value) OVER (
  [PARTITION BY partition_column]
  ORDER BY order_column
)
```



````python
%%sql
CREATE TABLE Sales (
  SaleDate DATE NOT NULL,
  Amount INT NOT NULL
);
INSERT INTO Sales (SaleDate, Amount)
VALUES
  ('2023-01-01', 100),
  ('2023-01-02', 150),
  ('2023-01-03', 200);
````

````python
%%sql
select * from Sales
````

````python
%%sql
SELECT
  SaleDate,
  Amount,
  LAG(Amount, 1, 0) OVER (ORDER BY SaleDate) AS PreviousAmount
FROM Sales;
````

````python
%%sql
SELECT
  SaleDate,
  Amount,
  LEAD(Amount, 1, 0) OVER (ORDER BY SaleDate) AS NextAmount
FROM Sales;
````
